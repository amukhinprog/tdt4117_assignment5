## 4.1 – Load and parse blog documents

Load all XML files from `blogs-sample`, extract the text of each `<post>`, and store one string per document. Keep the raw text per post as the document text.

You may experience some trouble parsing all lines in the files, but this is okay.



```python

# TODO: Load and parse the blog posts into a list named `documents`.

# Your code here
import os
import re
import html
import xml.etree.ElementTree as ET

FILE_PATH = r"./blogs-sample/" 

def parse_posts(path):
    parsed_files = []
    if not os.path.isdir(path):
        raise ValueError("Invalid folder path {path}")

    for filename in os.listdir(path):
        if filename.lower().endswith(".xml"):
            file_path = os.path.join(path, filename)
            try:
                with open(file_path, "r", encoding="latin-1") as f:
                    raw_content = f.read()
                content = raw_content.replace("&amp;", "___AMP___")

                content = html.unescape(content)

                content = content.replace("___AMP___", "&amp;")
                xml_content = re.sub(r'&(?!(#[0-9]+|[a-zA-Z0-9]+);)', '&amp;', content)

                root = ET.fromstring(xml_content)
                parsed_files.append((filename, root))
                #print(f"Successfully parsed: {filename}")
            except ET.ParseError as e:
                print(f"XML parsing error in {filename}: {e}")
            except Exception as e:
                print(f"Error reading {filename}: {e}")

    return parsed_files

def process_xml_data(parsed_files):
    doc_counter = 0
    all_posts_content = {}
    for filename, root in parsed_files:
        for post in root.findall('post'):
            if post.text:
                post_text = post.text.strip()
                all_posts_content[doc_counter] = post_text
                
        doc_counter += 1

    return all_posts_content

raw_documents = parse_posts(FILE_PATH)  

try:
    if raw_documents:
        docs = process_xml_data(raw_documents)
        print(docs[16])
    else:
        print("No XML files found in the folder.")
except ValueError as ve:
    print(ve)
#I have a problem parsing 24336.male.24.Technology.Leo.xml
```
